# Git Fundamentals

Git is a distributed version control system. It allows you to track changes in your code, revert to previous versions if something breaks, and collaborate with others without overwriting each other's work. 

Think of Git as a time machine and a highly organized filing system for your code. It tracks *who* made *what* changes and *when* they made them.

- **Working Directory**: The actual files you see and edit on your computer.
- **Staging Area**: A "waiting room" where you gather specific changes you are about to save.
- **Repository (Commit History)**: The permanent record of all saved versions of your project.

---

# 1. Initializing a Repository and Checking Status
To start using Git in a project, you must first initialize it. This creates a hidden `.git` folder that tracks all your changes. The `status` command is your best friend—it tells you exactly what is going on at any given moment.

```bash
# Navigate to your project folder
cd my_project

# Initialize a new Git repository
git init

# Check the current state of your files (untracked, modified, or staged)
git status
```

# 2. Staging Changes (git add)
When you create or modify files, Git notices them, but it doesn't save them automatically. You must explicitly move them to the Staging Area.

```bash
# Stage a specific file
git add main.py

# Stage a specific folder
git add data/

# Stage ALL modified and new files in the current directory
git add .
```

# 3. Saving Changes (git commit)
Once your changes are in the staging area, you "commit" them. A commit is a permanent snapshot of your project at that exact point in time. Every commit requires a brief, descriptive message.

```bash
# Commit the staged changes with a descriptive message
git commit -m "Add data preprocessing script"

# If you want to stage ALL tracked files AND commit in one step:
git commit -am "Fix bug in data loader"
```

# 4. Viewing History
You can always look back in time to see the history of your project, viewing every commit that has been made.

```bash
# View the full commit history (shows author, date, and message)
git log

# View a condensed, one-line version of the history
git log --oneline
```

# 5. Branching and Merging
Branches allow you to work on new features or experiment without affecting the main, stable version of your code (usually called `main` or `master`). Once your experiment is successful, you can merge it back into the main branch.

```bash
# List all branches (the active one has an asterisk *)
git branch

# Create a new branch called 'feature-login'
git branch feature-login

# Switch to the new branch
git checkout feature-login
# (Note: newer versions of Git also use 'git switch feature-login')

# --- Make some changes, add, and commit them on this branch ---

# Switch back to the main branch
git checkout main

# Merge the new feature into the main branch
git merge feature-login
```

## Real-World Use Case or Analogy:
Think of Git like playing a complex **Video Game**:

- **Working Directory**: You are actively running around, playing the level, fighting bosses, and losing health.
- **`git add` (Staging Area)**: You walk up to a save point. You haven't saved yet, but you are preparing to. You are gathering your inventory and deciding exactly what state you want to record.
- **`git commit`**: You press "Save Game". You now have a permanent save slot that you can load back into anytime you want. The commit message (`-m`) is the name you give that save slot (e.g., *"Beat the ice dragon, low health"*).
- **Branching (`git branch`)**: You are at a crossroads in the game and want to see what happens if you choose the evil path. You create a new "Profile" or "Save State" (branch) and play that storyline. Your original "good path" is safe and untouched. 
- **Merging (`git merge`)**: You discover the "evil path" gave you an amazing sword. You magically combine that sword back into your main profile so you can keep it forever.
